## 1. From a Python list to a NumPy array
`np.array()` turns a list into an `ndarray`: fixed type, fast vectorized math, any number of dimensions.

```python
mylist = [1,2,3]
```

```python
type(mylist)
```

```python
myarray = np.array(mylist)
```

```python
myarray
```

```python
type(myarray)
```

## 2. Built-in array creators
- `np.arange(start, stop, step)` → evenly spaced values, `stop` is **excluded**
- `np.zeros(shape)` / `np.ones(shape)` → arrays filled with 0 / 1 (default dtype is **float64**, that's why we see `0.`)

```python
np.arange(0, 10, 2)
```

```python
np.zeros(shape=(10,5))
```

> `0` is an `int`, `0.0` and `0.` are `float`. NumPy's `zeros`/`ones` return floats by default; pass `dtype=np.uint8` for image-like arrays.

```python
type(0)
```

```python
type(0.0)
```

```python
type(0.)
```

```python
np.ones(shape=(2,4))
# shape = (rows, columns)
# first value  -> number of rows    = height of the array
# second value -> number of columns = width of the array
```

## 3. Random integers and basic statistics
`np.random.seed(101)` makes the random numbers reproducible: same seed → same numbers.
`randint(low, high, size)` → `high` is **excluded**.
Note: `arr2` is different from `arr` because the seed is set only once; the generator keeps moving forward.

```python
np.random.seed(101)
arr = np.random.randint(0, 100, 10)
```

```python
arr
```

```python
arr2 = np.random.randint(0, 100, 10)
```

```python
arr2
```

- `max()` / `min()` → the value
- `argmax()` / `argmin()` → the **index** (position) of that value
- `mean()` → average

```python
arr.max()
```

```python
arr.argmax()
```

```python
arr.min()
```

```python
arr.argmin()
```

```python
arr.mean()
```

`reshape()` changes the shape without changing the data. The total number of elements must stay the same (10 = 2 × 5).

```python
arr.reshape((2,5))
```

## 4. Indexing and slicing a 2D array (a matrix)
`mat[row, col]` — rows first, then columns. This is exactly how we will access pixels: `img[y, x]`.

```python
mat = np.arange(0, 100).reshape(10,10)
```

```python
mat
```

```python
row = 0
col = 1
```

```python
mat[row, col]
```

- `mat[:, 1]` → every row, column 1 → a 1D array of shape `(10,)`
- `reshape(10, 1)` turns it into a real column vector of shape `(10, 1)`
- `mat[2, :]` → row 2, every column

```python
mat[:, 1]
```

```python
mat[:, 1].shape
```

```python
mat[:, 1].reshape(10,1)
```

```python
mat[2, :]
```

## 5. Assigning to a slice — and why `.copy()` matters
`mat[0:3, 0:3] = 0` changes the original `mat` in place (a slice is a **view**, not a copy).
To experiment safely, make an independent copy first with `.copy()`.

```python
mat[0:3, 0:3] = 0
```

```python
mat
```

```python
mynewmat = mat.copy()
```

```python
mynewmat
```

```python
mynewmat[0:6, :] = 99
```

```python
mynewmat
```
